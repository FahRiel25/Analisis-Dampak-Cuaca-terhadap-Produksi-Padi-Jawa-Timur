# Pipeline Data Cuaca

```python
# Baca semua data cuaca dalam folder
cuaca_files = glob.glob('/content/cuaca/*.csv')

list_cuaca = []

for file in cuaca_files:

    df = pd.read_csv(file, skiprows=3)

    # Ambil nama wilayah dari nama file
    wilayah = os.path.splitext(os.path.basename(file))[0]

    df['Wilayah'] = wilayah

    list_cuaca.append(df)

# Gabungkan seluruh file
df_cuaca_raw = pd.concat(list_cuaca, ignore_index=True)

df_cuaca_raw.head()
```

Cleaning

```python
# mengubah kolom time dari tipe data object/string menjadi format datetime
df_cuaca_raw['time'] = pd.to_datetime(df_cuaca_raw['time'])
```

```python
# Mengambil tahun dari kolom time kemudian disimpan kedalam kolom baru bernama tahun
df_cuaca_raw['Tahun'] = df_cuaca_raw['time'].dt.year

# Mengambil bulan dari kolom time kemudian disimpan kedalam kolom baru bernama bulan
df_cuaca_raw['Bulan'] = df_cuaca_raw['time'].dt.month
df_cuaca_raw.head()
```

```python
# Mengubah isi kolom Bulan dari angka menjadi Alfabet
bulan_mapping = {
    1: 'Januari',
    2: 'Februari',
    3: 'Maret',
    4: 'April',
    5: 'Mei',
    6: 'Juni',
    7: 'Juli',
    8: 'Agustus',
    9: 'September',
    10: 'Oktober',
    11: 'November',
    12: 'Desember'
}

df_cuaca_raw['Bulan'] = df_cuaca_raw['Bulan'].map(bulan_mapping)

df_cuaca_raw.head()
```

```python
# Mengubah Frekuensi waktu data cuaca dari harian menjadi bulanan
df_cuaca_bulanan = df_cuaca_raw.groupby(
    ['Wilayah', 'Tahun', 'Bulan']
).agg({
    'temperature_2m_mean (°C)': 'mean',
    'precipitation_sum (mm)': 'sum',
    'relative_humidity_2m_mean (%)': 'mean'
}).reset_index()

df_cuaca_bulanan.head()
```

```python
# Mengubah nama kolom agar mudah dipahami
df_cuaca_bulanan.rename(columns={
    'temperature_2m_mean (°C)': 'Rata_Rata_Suhu',
    'precipitation_sum (mm)': 'Total_Curah_Hujan',
    'relative_humidity_2m_mean (%)': 'Rata_Rata_Kelembapan'
}, inplace=True)

df_cuaca_bulanan.head()
```

```python
# Membuat kolom baru Urutan bulan agar bulan bisa urut Dari Januari Samapi Desember
urutan_bulan = {
    'Januari': 1,
    'Februari': 2,
    'Maret': 3,
    'April': 4,
    'Mei': 5,
    'Juni': 6,
    'Juli': 7,
    'Agustus': 8,
    'September': 9,
    'Oktober': 10,
    'November': 11,
    'Desember': 12
}

df_cuaca_bulanan['Urutan_Bulan'] = df_cuaca_bulanan['Bulan'].map(urutan_bulan)

# Data diurutkan berdasarkan Wilayah terlebih dahulu.Di dalam setiap wilayah, data diurutkan berdasarkan Tahun. Di dalam setiap tahun, data diurutkan berdasarkan Urutan_Bulan.
df_cuaca_bulanan = df_cuaca_bulanan.sort_values(
    by=['Wilayah', 'Tahun', 'Urutan_Bulan']
)
df_cuaca_bulanan.head()
```

```python
df_cuaca_bulanan.drop(columns='Urutan_Bulan', inplace=True)

df_cuaca_bulanan.head()
```

# Pipeline Data Produksi Padi

```python
# Baca data produksi padi
df_padi_raw = pd.read_csv('/content/padi/Produksi Padi Menurut KabupatenKota (Ton).csv')
df_padi_raw.head(10)
```

Transformasi format data dari Wide ke Long

```python
# mengambil tahun dan bulan dari header data BPS agar nantinya bisa digabung menjadi nama kolom yang lengkap
tahun = df_padi_raw.iloc[0].ffill()
bulan = df_padi_raw.iloc[1]
```

```python
# Membuat list untuk menyimpan nama kolom baru
new_columns = []

# Melakukan perulangan pada seluruh kolom dataset
for i in range(len(df_padi_raw.columns)):

    # Mengubah nama kolom pertama menjadi Wilayah
    if i == 0:
        new_columns.append('Wilayah')

    # Menggabungkan nama bulan dan tahun sebagai nama kolom
    else:
        col_name = f"{bulan[i]}_{int(tahun[i])}"
        new_columns.append(col_name)
```

```python
# Mengambil data produksi padi dan menghapus baris header yang tidak diperlukan
df_padi = df_padi_raw.iloc[2:].copy()

# Mengganti nama kolom dengan nama kolom yang telah dibuat sebelumnya
df_padi.columns = new_columns

# Menampilkan 5 baris pertama dataset
df_padi.head()
```

Menghapus kolom Serta baris yang tidak diperlukan

```python
print("Jumlah kolom baru:", len(new_columns))
```

```python
print(new_columns)
```

```python
kolom_tahunan = [
    col for col in df_padi.columns
    if 'Tahunan' in col
]

print(kolom_tahunan)
```

```python
df_padi.drop(
    columns=kolom_tahunan,
    inplace=True
)
```

```python
print(len(df_padi.columns))
```

```python
df_padi['Wilayah'].unique()
```

```python
# Menghapus baris Jawa Timur dan Wilayah sebelum dilakukan Melt
df_padi = df_padi[
    ~df_padi['Wilayah'].isin(
        ['Jawa Timur', 'Catatan']
    )
]
```

```python
# Menghapus baris yang kosong didalam kolom wilayah
df_padi = df_padi.dropna(
    subset=['Wilayah']
)
```

```python
df_padi['Wilayah'].unique()
```

```python
# Mengubah data dari format lebar (wide) menjadi format panjang (long)
df_padi_long = pd.melt(
    df_padi,
    id_vars=['Wilayah'],
    var_name='Periode',
    value_name='Produksi_Padi'
)

# Menampilkan 5 baris pertama data
df_padi_long.head()
```

```python
# Memisahkan informasi bulan dan tahun dari kolom Periode ke kolom baru
df_padi_long[['Bulan', 'Tahun']] = df_padi_long['Periode'].str.split('_', expand=True)
```

```python
# Mengatur urutan kolom sesuai kebutuhan analisis
df_padi_long = df_padi_long[
    ['Wilayah', 'Tahun', 'Bulan', 'Produksi_Padi']
]

# Menampilkan 5 baris pertama data
df_padi_long.head()
```

```python
df_padi_long.info()
```

```python
# Konversi tipe data tahun dan produksi padi menjadi numeric
df_padi_long['Tahun'] = pd.to_numeric(
    df_padi_long['Tahun']
)

df_padi_long['Produksi_Padi'] = pd.to_numeric(
    df_padi_long['Produksi_Padi'],
    errors='coerce'
)
```

# Rolling 4 Bulan data cuaca

```python
urutan_bulan = {
    'Januari':1,
    'Februari':2,
    'Maret':3,
    'April':4,
    'Mei':5,
    'Juni':6,
    'Juli':7,
    'Agustus':8,
    'September':9,
    'Oktober':10,
    'November':11,
    'Desember':12
}

df_final['Bulan_Angka'] = df_final['Bulan'].map(urutan_bulan)
df_final.head()
```

```python
df_final = df_final.sort_values(
    ['Wilayah', 'Tahun', 'Bulan_Angka']
)
```

```python
df_final[
    df_final['Wilayah'] == 'Kabupaten Pacitan'
][['Tahun','Bulan','Bulan_Angka']].head(15)
```

Rolling 4 bulan

```python
df_final['Suhu_4Bln'] = (
    df_final
    .groupby('Wilayah')['Rata_Rata_Suhu']
    .transform(
        lambda x: x.rolling(
            window=4,
            min_periods=4
        ).mean()
    )
)
```

```python
df_final['Curah_4Bln'] = (
    df_final
    .groupby('Wilayah')['Total_Curah_Hujan']
    .transform(
        lambda x: x.rolling(
            window=4,
            min_periods=4
        ).sum()
    )
)
```

```python
df_final['Kelembapan_4Bln'] = (
    df_final
    .groupby('Wilayah')['Rata_Rata_Kelembapan']
    .transform(
        lambda x: x.rolling(
            window=4,
            min_periods=4
        ).mean()
    )
)
```

```python
df_final[
    df_final['Wilayah'] == 'Kabupaten Pacitan'
][[
    'Tahun',
    'Bulan',
    'Rata_Rata_Suhu',
    'Suhu_4Bln',
    'Total_Curah_Hujan',
    'Curah_4Bln',
    'Rata_Rata_Kelembapan',
    'Kelembapan_4Bln'
]].head(10)
```

```python
df_final[
    [
        'Suhu_4Bln',
        'Curah_4Bln',
        'Kelembapan_4Bln'
    ]
].isna().sum()
```

```python
# Menghapus baris bulan yang bernilai NaN dikarenakan tidak ada data historis cuaca pada bulan sebelumnya
df_final = df_final.dropna(
    subset=[
        'Suhu_4Bln',
        'Curah_4Bln',
        'Kelembapan_4Bln'
    ]
)
```
